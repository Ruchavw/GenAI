# RAG Assignment

## Objective
To build a simple RAG (Retrieval-Augmented Generation) system using LangChain and FAISS.

---

## What is RAG?
RAG is a method where:
- we retrieve relevant data from documents
- then use an LLM to generate answers based on that data

---

## Steps Done

1. Took a sample text about AI  
2. Split the text into chunks  
3. Converted chunks into embeddings  
4. Stored them using FAISS  
5. Created a retriever  
6. Connected it with an LLM (Groq)  
7. Built a pipeline to answer questions  

---

## Flow

Question → Retriever → Context → LLM → Answer

---

## Sample Questions & Answers

**Q:** What is AI?  
**A:** AI is the simulation of human intelligence in machines.

**Q:** What is machine learning?  
**A:** It is a subset of AI that learns from data.

**Q:** Where is AI used?  
**A:** Healthcare, finance, vehicles, chatbots.

---

## Key Points

- FAISS is used for fast search  
- Chunking helps in better retrieval  
- Embeddings convert text to vectors

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import FakeEmbeddings

text = """Artificial Intelligence (AI) is the simulation of human intelligence in machines.
It enables computers to perform tasks such as learning, reasoning, and problem-solving.
AI is used in applications like healthcare, finance, autonomous vehicles, and chatbots.
Machine learning is a subset of AI that allows systems to learn from data."""

splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=50)
docs = splitter.create_documents([text])

embedding = FakeEmbeddings(size=384)

vectorstore = FAISS.from_documents(docs, embedding)
retriever = vectorstore.as_retriever()

print("WORKING")
```

```python
from langchain_groq import ChatGroq

GROQ_API_KEY = "your_key"

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.5,
    groq_api_key=GROQ_API_KEY
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
Answer the question using only the context below:
{context}

Question: {input}
""")
```

```python
def rag_pipeline(question):
    # Retrieve docs
    docs = retriever.invoke(question)
    # Combine context
    context = "\n".join([doc.page_content for doc in docs])

    # Format prompt
    formatted_prompt = prompt.format(context=context, input=question)

    # Call LLM
    response = llm.invoke(formatted_prompt)

    return response.content
```

```python
print(rag_pipeline("What is this document about?"))
```

```python
test_questions = [
    "What is Artificial Intelligence?",
    "What is machine learning?",
    "Where is AI used?",
    "Summarize the document"
]

for q in test_questions:
    print("\nQ:", q)
    print("A:", rag_pipeline(q))
```
