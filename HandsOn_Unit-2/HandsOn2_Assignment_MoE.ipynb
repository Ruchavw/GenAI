```python
from dotenv import load_dotenv
load_dotenv()

import getpass
import os
from langchain_groq import ChatGroq

if "GROQ_API_KEY" not in os.environ:
    os.environ["GROQ_API_KEY"] = getpass.getpass("Enter your Groq API Key: ")

llm_router = ChatGroq(model="llama-3.1-8b-instant", temperature=0.0)
```

```python
llm_technical = ChatGroq(model="llama-3.1-8b-instant", temperature=0.7)

llm_billing = ChatGroq(model="llama-3.1-8b-instant", temperature=0.7)

llm_general = ChatGroq(model="llama-3.1-8b-instant", temperature=0.7)
```

```python
SYSTEM_PROMPTS = {
    "technical": (
        "You are a senior software engineer. "
        "Be precise, technical, and code-focused. "
        "Explain errors clearly and suggest fixes."
    ),
    "billing": (
        "You are a customer support billing expert. "
        "Be empathetic, policy-driven, and explain refunds "
        "and subscription issues clearly."
    ),
    "general": (
        "You are a friendly AI assistant for general conversation."
    )
}
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

def route_prompt(user_input: str) -> str:
    router_prompt = (
        "Classify the following user request into one of these categories:\n"
        "[technical, billing, general, tool]\n\n"
        "Return ONLY the category name.\n\n"
        "If the request is about 'bitcoin price' classify it as 'tool'.\n\n"
        f"User request: {user_input}"
    )

    messages = [
        SystemMessage(content="You are an intent classification engine."),
        HumanMessage(content=router_prompt)
    ]

    response = llm_router.invoke(messages)
    return response.content.strip().lower()
```

```python
def fetch_bitcoin_price():
    return "Bitcoin price today is approximately $43,000 USD."
```

```python
def process_request(user_input: str) -> str:
    category = route_prompt(user_input)
    print(f"[Router Decision] → {category}")

    if category == "tool":
        if "bitcoin" in user_input.lower():
            return fetch_bitcoin_price()
        else:
            return "Requested tool is not supported."

    if category == "technical":
        llm = llm_technical
    elif category == "billing":
        llm = llm_billing
    else:
        llm = llm_general
        category = "general"

    messages = [
        SystemMessage(content=SYSTEM_PROMPTS[category]),
        HumanMessage(content=user_input)
    ]

    response = llm.invoke(messages)
    return response.content
```

```python
query = "My Python script throws an IndexError on line 5."
print(process_request(query))
```

```python
query = "I was charged twice for my subscription this month."
print(process_request(query))
```
